```python
# while loop for 3d
import numpy as np
arr = np.arange(24).reshape(2, 3, 4)


i = 0
while i < len(arr):
  j = 0
  while j < len(arr[i]):
    k = 0
    while k < len(arr[i][j]):
      print(arr[i][j][k], end=" ")
      k += 1
    print()
    j += 1
  print()
  i += 1
```

```python
#for loop for 3d
import numpy as np
arr = np.arange(24).reshape(2, 3, 4)
for i in arr:
    for j in i:
        for k in j:
            print(k, end=" ")
        print()
    print()
```

```python
import numpy as np
# for loop 1d array
arr2=np.arange(12)
for i in range(len(arr2)):
    print(arr2[i],end=" ")
print("\n")
# for loop 2d array
arr3=np.arange(12).reshape(3,4)
for i in range(len(arr3)):
    for j in range(len(arr3[i])):
        print(arr3[i][j],end=" ")
    print()
print("\n")
# for loop 3d array
arr4=np.arange(12).reshape(2,2,3)
# print(arr4)
for i in range(len(arr4)):
    print(f"layer {i}")
    for j in range(len(arr4[i])):
        for k in range(len(arr4[i][j])):
            print(arr4[i][j][k],end=" ")
        print()
    print()
```
