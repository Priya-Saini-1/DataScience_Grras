Relational operator
==, < , >, >=, <=, !=

```python
num1 = 10
num2 = 11

print(num1 == num2)
print(num1 < num2)
print(num1 > num2)
print(num1 <= num2)
print(num1 >= num2)
print(num1 != num2)
```

```python
#Assignment operator
# = , += , -= , *= , /= , //= , %=

# =
a = 10
b = a + 20
print(b)

# +=
a = 10
a += 20
print(a)

# -=
a = 10
a -= 20
print(a)

# *=
a = 10
a *= 20
print(a)


# /=
a = 10
a /= 20
print(a)

# //=
a = 10
a //= 20
print(a)
```
