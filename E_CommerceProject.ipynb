***E-Commerce Sales Analysis using Python, NumPy, Pandas and Matplotlib***

Importing Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

Creating Dataset (for E-commerce)

```python
data = {
    "OrderID":[1001,1002,1003,1004,1005,1006,1007,1008,1009,1010],
    "Product":["Laptop","Mouse","T-Shirt","Shoes","Mobile",
               "Watch","Bag","Headphones","Keyboard","Tablet"],
    "Category":["Electronics","Electronics","Fashion","Fashion",
                "Electronics","Accessories","Fashion",
                "Electronics","Electronics","Electronics"],
    "Price":[50000,500,800,2500,20000,1500,1200,2500,1200,18000],
    "Quantity":[2,5,4,3,2,6,5,4,3,2],
    "Month":["Jan","Jan","Feb","Feb","Mar",
             "Mar","Apr","Apr","May","May"]
}

df = pd.DataFrame(data)

print(df)
```

Creating Revenue column

```python
df["Revenue"] = df["Price"] * df["Quantity"]

print(df)
```

Basic Dataset Info

```python
print("Dataset Info\n")
print(df.info())

print("\nDataset Description\n")
print(df.describe())
```

Total Revenue

```python
total_revenue = np.sum(df["Revenue"])

print("Total Revenue = ₹", total_revenue)
```

Average Revenue

```python
average_revenue = np.mean(df["Revenue"])

print("Average Revenue = ₹", average_revenue)
```

Maximum Revenue

```python
max_revenue = np.max(df["Revenue"])

print("Maximum Revenue = ₹", max_revenue)
```

Minimum Revenue

```python
min_revenue = np.min(df["Revenue"])

print("Minimum Revenue = ₹", min_revenue)
```

Standard Deviation

```python
std_revenue = np.std(df["Revenue"])

print("Standard Deviation =", std_revenue)
```

Top Selling Product

```python
top_selling = df.groupby("Product")["Quantity"].sum()

print(top_selling)

print("\nTop Selling Product")

print(top_selling.idxmax())
```

Most Profitable Product

```python
profit_product = df.groupby("Product")["Revenue"].sum()

print(profit_product)

print("\nMost Profitable Product")

print(profit_product.idxmax())
```

Category wise Revenue

```python
category_revenue = df.groupby("Category")["Revenue"].sum()

print(category_revenue)
```

Monthly Revenue

```python
monthly_revenue = df.groupby("Month")["Revenue"].sum()

print(monthly_revenue)
```

Revenue by Category (Bar Chart)

```python
category_revenue.plot(kind="bar")

plt.title("Revenue by Category")
plt.xlabel("Category")
plt.ylabel("Revenue")

plt.show()
```

Monthly Revenue Trend (Line Chart)

```python
monthly_revenue.plot(kind="line", marker="o")

plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue")

plt.show()
```

Category Contribution (Pie chart)

```python
category_revenue.plot(
    kind="pie",
    autopct="%1.1f%%"
)

plt.title("Category Contribution")

plt.ylabel("")

plt.show()
```

Top 5 products by Revenue

```python
top_products = df.groupby("Product")["Revenue"].sum()

top_products = top_products.sort_values(
    ascending=False
)

print(top_products.head())
```

Products with Revenue > ₹10000

```python
high_revenue = df[df["Revenue"] > 10000]

print(high_revenue)
```
