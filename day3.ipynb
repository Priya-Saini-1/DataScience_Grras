```python
#Lambda functions
add = lambda x: x
print(add(10))

sum = lambda x, y: x+y
print(sum(10, 20))

a = lambda *x : [i  for i in x]
print(a(10, 20, 30, 40, 50))
```

```python
#list comprehensions [expression for item in iterable]
print([i for i in range(5)])

# example
l = [10, 20, 30, 40, 50]
print([l[i] for i in range(len(l))])

#question
l = [10, 20, [30, 40, 50, 60]]

print([l[-1][i] for i in range(len(l[-1]))])
```
