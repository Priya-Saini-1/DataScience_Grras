Slicing

```python
# nested list
l = [10, 20, 30, ["hello", "hello1", "hello2", [True, False]]]
for i in range(len(l)):
    if type(l[i]) == list:
        for j in range(len(l[i])):
            if type(l[i][j]) == list:
                for k in range(len(l[i][j])):
                    print(l[i][j][k])
            else:
                print(l[i][j])
    else:
        print(l[i])

# slicing
l1 = l[0:4]
print(l1[1:3])

print(l1[:3])

# example
print(l[2:])
```
